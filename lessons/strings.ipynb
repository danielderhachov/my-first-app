# Підстановка значень у стрічки

```python
my_account = 1000
message = 'I managed to get %s points'
print(message % my_account)
```

```python
жарт='%s: пристрій для знаходження меблів у темряві'
а1='коліно'
а2='гомілка'
print(жарт % а1)
print(жарт % а2)
```

```python
числа='що число %s сказало числу %s? - Гарний пояс!'
print(числа % (0, 8))
```
